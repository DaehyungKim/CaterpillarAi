```python
import commentjson as json  

with open("entities/titles_short.jsonc", "r", encoding="utf-8") as f:
    data = json.load(f)  


unique_data = []
seen = set()

for item in data:
    key = json.dumps(item, sort_keys=True)
    if key not in seen:
        seen.add(key)
        unique_data.append(item)

result = json.dumps(unique_data, indent=2, ensure_ascii=False)
print(result)
```
